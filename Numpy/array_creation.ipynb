```python
import numpy as np
```

```python
# Create a 1D Array mannually 
arr = np.array([1,2,3,4,5])
arr
```

```python
# dtype
arr = np.array([1,2,3,4.5,12.3], dtype=int)
arr
```

```python
# Create array from lists
li = [12,56,78,45,9,6]
arr = np.array(li)
arr
```

```python
arr.shape
```

```python
# Create a 2D array from 2D list
li = [[1,2],[3,4],[5,6]]
arr = np.array(li)
arr
```

```python
# Create 2D array mannually
arr2D = np.array([(1,2,3),(4,5,6)])
arr2D
```

```python
arr2D.shape
```

```python
# Create 3D array from list 
li = [[[1,2],[3,4]],[[5,6],[7,8]]]
arr3D = np.array(li)
arr3D
```
