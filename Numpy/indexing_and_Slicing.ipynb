Numpy Array Indexing

```python
import numpy as np
```

```python
# 1D array
arr = np.arange(1,11,1)
arr 
```

```python
arr[4]
arr[9]
arr[0]
```

```python
arr = np.random.randint(1,20,(3,3))
arr 
```

```python
arr[0,2]
```

Numpy Array Slicing 

```python
arr = np.arange(6)
arr 
```

```python
# Slice a specific range " : "
arr[2:5]
```

```python
arr[ : ]
```

```python
arr[2: ]
```

```python
arr[ : 2]
```

```python
# Reverse a array
arr[ :: -1]
```

```python
arr
```

```python
arr[ :: 2]
```

```python
arr = np.random.randint(1,30,(5,5))
arr 
```

```python
# row s:e , c s:e
arr[ : 2, : 2]
```

```python
arr[2: ,2: ]
```

```python
arr[ : , : ]
```

```python
arr[2:4, : 3]
```

```python
# Skipping rows  
arr[::2, :]
```
