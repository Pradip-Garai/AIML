Append

```python
import numpy as np
```

```python
# 1d Array
arr = np.arange(1,5)
arr 
```

```python
# add a new element
new = np.append(arr,[6,7,8])
new 
```

```python
# 2d array
arr2D = np.random.randint(1,50,(4,5))
arr2D 
```

```python
# insert at row 
new = np.append(arr2D,[[12,36,56,89,52]], axis=0)
new 
```

```python
# insert at column
new = np.append(arr2D, [[100],[200],[300],[400]], axis=1)
new 
```

Insert at specific position

```python
arr = np.arange(1,5,1)
arr 
```

```python
new = np.insert(arr,2,[10,20])
new 
```

```python
# insert at 2d array
a = np.random.randint(1,50,(4,3))
a 
```

```python
# Insert at row 
new = np.insert(a,[1],20, axis=0)
new 
```

```python
# insert at column 
new = np.insert(a,1,[[100],[200],[300],[400]],axis=1)  
new 
```

```python
# Delete 
new = np.delete(a,1,axis=0)
new 
```
