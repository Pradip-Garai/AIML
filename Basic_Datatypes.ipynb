Basic Datatypes in Python

 1. Integer
 2. Floating
 3. Complex
 4. Boolean
 5. String

```python
# Integer
a = 8
print(a)
type(a)
```

```python
# Floating Point
b = 2.36
print(b)
type(b)
```

```python
# Complex Numbers
c = 1 + 3j
print(c)
type(c)
```

```python
# int to float conversion
x = 10
print(x)
type(x)
```

```python
y = float(x)
print(y)
type(y)
```

```python
# float to int conversion
x = 50.10
print(x)
type(x)
```

```python
y = int(x)
print(y)
type(y)
```
